```python
a1 = np.array([10,20,30,40])
a2 = np.array([[1,2,3],[4,5,6]])
a3 = np.arange(0,10,2)
a4 = np.linspace(0,1,5)
a5 = np.zeros((2,3))
a6 = np.ones((3,2))

for name, x in [('a1', a1), ('a2', a2), ('a3', a3), ('a4', a4), ('a5', a5), ('a6', a6)]:
    print(name, "| shape:", x.shape, "| dtype:", x.dtype, "| ndim:", x.ndim, "| size:", x.size)
```
